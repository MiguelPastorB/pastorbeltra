# 2. Desarrollo <a name="desarrollo"></a>

Desde una perspectiva formal, hemos modelado un entorno estacionario compuesto por una máquina de $k$ brazos, donde cada acción $a_t$ desencadena una recompensa $r_t$ regida por una distribución de probabilidad $\mathcal{P}_i$ inicialmente desconocida para el agente . El éxito de nuestra intervención se mide a través del "arrepentimiento" (regret), una métrica que cuantifica la pérdida acumulada por no haber seleccionado siempre la acción óptima $\mu^*$:

$$R(T) = T\mu^* - E \left[ \sum_{t=1}^{T} \mu_{a_t} \right]$$

Para abordar este problema, hemos seleccionado cuatro enfoques, como son los métodos $\epsilon$-greedy, $\epsilon$-decaimiento, UCB1 y Softmax. Hemos diseñado experimentos que ponen a prueba la sensibilidad de estos algoritmos ante cambios en el número de ensayos ($n$) o en el ruido del entorno ($\sigma$), garantizando siempre la reproducibilidad mediante el uso de semillas fijas en todas las ejecuciones.

# 4. Conclusiones <a name="conclusiones"></a>

A pesar de que cada notebook tiene sus propias conclusines más detalladas, dejamos una breve comparativa entre ellos:

| Entorno de Recompensa | Algoritmo Ganador | Análisis |
| :--- | :--- | :--- |
| **Bernoulli** ($n=1$) | **$\epsilon$-greedy** ($\epsilon=0.1$) | En entornos binarios estáticos, su simplicidad permite una convergencia más veloz que métodos basados en confianza[cite: 358, 359]. |
| **Binomial** ($n=10$) | **UCB1** ($c=1$) | Al procesar $n=10$ ensayos por paso, la incertidumbre sobre la media se reduce drásticamente, optimizando el uso de UCB1[cite: 395]. |
| **Normal** ($\sigma=1.0$) | **UCB1** ($c=1$) | En entornos continuos con bajo ruido, el optimismo ante la incertidumbre logra una selección óptima casi perfecta[cite: 399, 474]. |
| **Normal** ($\sigma=20.0$) | **$\epsilon$-decaimiento** | El ruido extremo "engaña" los intervalos de UCB1. Una exploración aleatoria que decrece resulta ser la opción más robusta[cite: 455, 459]. |

A partir de **nuestros ensayos**, hemos analizado el comportamiento de cada algoritmo según su respuesta al dilema de exploración-explotación:

* **$\epsilon$-greedy (Fijo):** hemos comprobado que es la opción más robusta frente al ruido. Sin embargo, su arrepentimiento acumulado siempre será lineal por su exploración residual constante, lo que supone una ineficiencia en horizontes largos.
* **$\epsilon$-decaimiento:** consideramos que es el método con mayor potencial a largo plazo. Al reducir gradualmente la exploración, logra el equilibrio más fino entre aprendizaje inicial y aprovechamiento de la recompensa final.
* **UCB1:** hemos comprobado que es imbatible en entornos con alta relación señal-ruido. No obstante, su rendimiento decae drásticamente si el ruido ($\sigma$) es comparable a la diferencia entre las medias de los brazos.
* **Softmax:** confirmamos que es el algoritmo más sensible a la escala de las recompensas. Requiere una sintonización crítica de la temperatura ($\tau$) para evitar comportamientos puramente aleatorios o prematuramente codiciosos.
