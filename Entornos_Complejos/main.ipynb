## 4.2. Conclusiones generales sobre los agentes

* **Agente Monte Carlo (Tabular):** hemos comprobado que este agente es el más dependiente de la estructura del entorno. Debido a que requiere completar el episodio para aprender, sufre enormemente en entornos de recompensa dispersa (como FrozenLake), donde su incapacidad para realizar bootstrapping lo deja sin información útil durante gran parte del entrenamiento.

* **Agente Q-Learning (Tabular):** es nuestro agente más ambicioso. Al ser un algoritmo off-policy, ignora las acciones de exploración para actualizar sus valores Q basándose siempre en la mejor acción posible. Esto lo hace extremadamente eficiente encontrando la ruta óptima, aunque en entornos como CliffWalking, esta valentía le lleva a caer al acantilado con mayor frecuencia durante el aprendizaje.

* **Agente SARSA (Tabular):** este agente ha demostrado una naturaleza más prudente y realista. Al ser on-policy, aprende los valores de la política que realmente está ejecutando (incluyendo sus errores de exploración). Esto lo convierte en nuestra opción predilecta para entornos con penalizaciones severas, ya que aprende a alejarse de los peligros, sacrificando la ruta más corta por la más segura.

* **Agente DQN (Approximate):** hemos verificado que su éxito en espacios continuos depende de dos componentes críticos: el Replay Buffer, que le permite reutilizar experiencias pasadas de forma aleatoria, y la Target Network, que estabiliza el entrenamiento al proporcionar objetivos de aprendizaje fijos. Sin estos mecanismos, el agente sería incapaz de equilibrar el poste en CartPole debido a la inestabilidad de las redes neuronales.

* **Agente SARSA Semi-gradiente (Approximate):** en nuestra comparativa de aproximación, este agente ha servido para ilustrar las dificultades de los métodos on-policy en Deep RL sin memoria. Al carecer de un buffer de repetición y aprender de la trayectoria inmediata, es mucho más propenso a la inestabilidad y presenta una convergencia más lenta, lo que refuerza nuestra preferencia por arquitecturas tipo DQN para control continuo complejo.

# 5. Uso de la IA generativa <a name="ia-generativa"></a>

### Prompt 1. ¿Por qué Monte Carlo muestra valores de recompensa tan bajos en CliffWalking comparado con Q-Learning?
Esto ocurre porque Monte Carlo espera a terminar el episodio para actualizar la tabla Q. En entornos con penalizaciones por paso y caídas, el agente acumula una cantidad masiva de recompensas negativas antes de realizar su primera actualización. A diferencia de Q-Learning, que aprende paso a paso mediante diferencias temporales, Monte Carlo sufre de una varianza extrema que "envenena" sus estimaciones de valor si no alcanza la meta rápidamente.

### Prompt 2. En FrozenLake estocástico, ¿qué diferencia práctica hay entre la actualización de SARSA y Q-Learning?

La diferencia reside en cómo gestionan el riesgo del "hielo resbaladizo". SARSA es on-policy y actualiza el valor considerando la acción que realmente va a tomar (que puede ser un error de exploración), lo que le lleva a aprender rutas más seguras. Q-Learning es off-policy y actualiza asumiendo que siempre tomará la mejor acción posible en el siguiente estado, lo que le permite encontrar la ruta óptima pero le hace caer más veces durante el entrenamiento debido a su naturaleza arriesgada.

### Prompt 3. ¿Por qué mi agente DQN tiene "caídas" repentinas en su rendimiento después de haber alcanzado la recompensa máxima?

Este fenómeno se conoce como "olvido catastrófico" o inestabilidad por sobreajuste. En entornos como CartPole, cuando el agente se vuelve muy bueno, el Replay Buffer se llena casi exclusivamente de experiencias de "éxito" (poste equilibrado). La red neuronal deja de ver ejemplos de fallos y "olvida" cómo reaccionar ante ángulos extremos. Para evitarlo, se recomienda usar un Replay Buffer más grande, ajustar el decaimiento de epsilon para mantener una exploración residual o emplear técnicas como Prioritized Experience Replay.
